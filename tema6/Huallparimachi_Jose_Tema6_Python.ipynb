```python
import numpy as np
import matplotlib.pyplot as plt

print("Entorno preparado correctamente")
print("Versión de NumPy:", np.__version__)
```

```python
# Crear un degradado horizontal de 5 valores entre 0 y 255
degradado = np.linspace(0, 255, 5, dtype=np.uint8)

# Repetir el degradado en 5 filas para formar una imagen de 5 x 5
gris = np.tile(degradado, (5, 1))

# Mostrar las características de la imagen
print("Matriz de la imagen:")
print(gris)

print("Dimensiones (shape):", gris.shape)
print("Tipo de dato (dtype):", gris.dtype)
print("Promedio de los píxeles:", gris.mean())
```

```python
# Visualizar la imagen en escala de grises
plt.imshow(gris, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen en escala de grises - Tema 6")
plt.colorbar(label="Intensidad del píxel")
plt.savefig("gris_tema6.png")
plt.show()
```

```python
# Recortar las últimas 3 columnas de la imagen
recorte = gris[:, 2:5]

print("Recorte de la mitad derecha:")
print(recorte)
print("Dimensiones del recorte:", recorte.shape)
```

```python
# Crear una imagen RGB de 4 filas, 4 columnas y 3 canales
rgb = np.zeros((4, 4, 3), dtype=np.uint8)

# Mitad izquierda: rojo (R=255, G=0, B=0)
rgb[:, 0:2] = [255, 0, 0]

# Mitad derecha: azul (R=0, G=0, B=255)
rgb[:, 2:4] = [0, 0, 255]

print("Dimensiones de la imagen RGB:", rgb.shape)
print("Pixel superior izquierdo:", rgb[0, 0])
print("Pixel superior derecho:", rgb[0, 3])

# Mostrar la imagen
plt.imshow(rgb)
plt.title("Imagen RGB: rojo y azul")
plt.show()
```

```python
# Extraer los tres canales de color
canal_R = rgb[:, :, 0]
canal_G = rgb[:, :, 1]
canal_B = rgb[:, :, 2]

print("Canal rojo (R):")
print(canal_R)

print("\nCanal verde (G):")
print(canal_G)

print("\nCanal azul (B):")
print(canal_B)
```

```python
# Conversión mediante promedio simple
gris_promedio = rgb.mean(axis=2)

# Conversión mediante luminosidad ponderada
R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

gris_luminosidad = 0.299 * R + 0.587 * G + 0.114 * B

print("Gris mediante promedio simple:")
print(gris_promedio)

print("\nGris mediante luminosidad ponderada:")
print(gris_luminosidad)
```
