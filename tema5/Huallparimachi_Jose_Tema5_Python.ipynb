```python
import pandas as pd

df = pd.read_csv("estudiantes_limpio.csv")

df
```

```python
import numpy as np

media = df["nota"].mean()
mediana = df["nota"].median()
desv_std = df["nota"].std()
varianza = np.var(df["nota"])
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print("Media:", media)
print("Mediana:", mediana)
print("Desviación estándar:", desv_std)
print("Varianza poblacional:", varianza)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
from scipy import stats

asimetria_pandas = df["nota"].skew()
curtosis_pandas = df["nota"].kurt()

asimetria_scipy = stats.skew(df["nota"])
curtosis_scipy = stats.kurtosis(df["nota"])

print("Asimetría (pandas):", asimetria_pandas)
print("Curtosis (pandas):", curtosis_pandas)
print("Asimetría (SciPy):", asimetria_scipy)
print("Curtosis (SciPy):", curtosis_scipy)
```

```python
def resumen_estadistico(serie, decimales=4):
    n = serie.count()
    media = serie.mean()
    mediana = serie.median()
    desv_std = serie.std()
    cv_pct = (desv_std / media) * 100

    return {
        "n": n,
        "media": round(media, decimales),
        "mediana": round(mediana, decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(cv_pct, decimales)
    }
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"
```

```python
for variable in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[variable])
    dispersion = clasificar_dispersion(resumen["cv_pct"])

    print("\nVariable:", variable)
    print("Resumen:", resumen)
    print("Dispersión:", dispersion)
```
